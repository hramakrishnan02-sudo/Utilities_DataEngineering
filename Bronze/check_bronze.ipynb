```python
bronze_path = "abfss://bronze@utilitiesmaintainence.dfs.core.windows.net/"
maintenance_df = (
    spark.read
         .option("header", True)
         .option("inferSchema", True)
         .csv(
             bronze_path + "monthly_files/maintenance_2026_01.csv"
         )
)

display(maintenance_df)
maintenance_df.printSchema()
```

```python
from pyspark.sql.functions import col, count

maintenance_df.filter(
    col("work_order_id").isNull()
).count()
```

```python
duplicates_df = (
    maintenance_df
    .groupBy("work_order_id")
    .count()
    .filter(col("count") > 1)
)

display(duplicates_df)
```

```python
for month in ["01", "02", "03"]:
    df = (
        spark.read
        .option("header", True)
        .option("inferSchema", True)
        .csv(
            bronze_path +
            f"monthly_files/maintenance_2026_{month}.csv"
        )
    )

    print(f"===== 2026_{month} =====")
    df.printSchema()
```

```python
from pyspark.sql.functions import col, lit

required_columns = [
    "work_order_id",
    "asset_id",
    "technician_id",
    "location_id",
    "maintenance_type",
    "maintenance_category",
    "planned_flag",
    "failure_type",
    "failure_severity",
    "priority",
    "work_order_status",
    "resolution_code",
    "created_date",
    "scheduled_date",
    "completed_date",
    "downtime_hours",
    "labor_hours",
    "parts_used",
    "parts_cost",
    "labor_cost",
    "total_cost",
    "source_system",
    "updated_at"
]

missing_columns = [
    c for c in required_columns
    if c not in maintenance_df.columns
]

extra_columns = [
    c for c in maintenance_df.columns
    if c not in required_columns
]

print("Missing columns:", missing_columns)
print("Extra columns:", extra_columns)
```

```python
from pyspark.sql.functions import col, sum, when

null_counts = maintenance_df.select(
    [
        sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
        for c in maintenance_df.columns
    ]
)

display(null_counts)
```

```python
from pyspark.sql.functions import col

display(
    maintenance_df
    .filter(col("completed_date").isNull())
    .groupBy("work_order_status")
    .count()
    .orderBy("work_order_status")
)
```

```python
from pyspark.sql.functions import col, trim, when

clean_df = maintenance_df

string_columns = [
    field.name
    for field in maintenance_df.schema.fields
    if field.dataType.simpleString() == "string"
]

for c in string_columns:
    clean_df = clean_df.withColumn(
        c,
        when(trim(col(c)) == "", None)
        .otherwise(trim(col(c)))
    )

display(clean_df)
```

```python
mandatory_columns = [
    "work_order_id",
    "asset_id",
    "technician_id",
    "location_id"
]

for c in mandatory_columns:
    print(
        c,
        clean_df.filter(col(c).isNull()).count()
    )
```

```python
duplicate_df = (
    clean_df
    .groupBy("work_order_id")
    .count()
    .filter(col("count") > 1)
)

display(duplicate_df)

print("Duplicate work_order_id groups:", duplicate_df.count())
```
