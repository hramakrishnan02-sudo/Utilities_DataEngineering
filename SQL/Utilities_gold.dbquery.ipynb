```sql
SELECT current_catalog(), current_schema();
SHOW TABLES IN motordatabricks.utilities_gold;

-- ============================================================
-- 1. Create Analytics Schema
-- ============================================================

CREATE SCHEMA IF NOT EXISTS motordatabricks.utilities_analytics;


-- ============================================================
-- 2. Maintenance Cost Summary
-- ============================================================

CREATE OR REPLACE VIEW motordatabricks.utilities_analytics.maintenance_cost_summary AS
SELECT
    a.asset_type,
    a.manufacturer,
    COUNT(*) AS maintenance_count,
    ROUND(SUM(f.total_cost), 2) AS total_maintenance_cost,
    ROUND(AVG(f.total_cost), 2) AS avg_maintenance_cost,
    ROUND(SUM(f.labor_cost), 2) AS total_labor_cost,
    ROUND(SUM(f.parts_cost), 2) AS total_parts_cost
FROM motordatabricks.utilities_gold.fact_maintenance f
JOIN motordatabricks.utilities_gold.dim_asset a
    ON f.asset_key = a.asset_key
GROUP BY
    a.asset_type,
    a.manufacturer;


-- ============================================================
-- 3. Maintenance By Asset
-- ============================================================

CREATE OR REPLACE VIEW motordatabricks.utilities_analytics.maintenance_by_asset AS
SELECT
    a.asset_id,
    a.asset_type,
    a.manufacturer,
    a.model,
    a.city,
    a.region,
    a.criticality,
    a.asset_status,
    COUNT(*) AS maintenance_count,
    ROUND(SUM(f.downtime_hours), 2) AS total_downtime_hours,
    ROUND(SUM(f.labor_hours), 2) AS total_labor_hours,
    ROUND(SUM(f.total_cost), 2) AS total_maintenance_cost,
    ROUND(AVG(f.total_cost), 2) AS avg_maintenance_cost
FROM motordatabricks.utilities_gold.fact_maintenance f
JOIN motordatabricks.utilities_gold.dim_asset a
    ON f.asset_key = a.asset_key
GROUP BY
    a.asset_id,
    a.asset_type,
    a.manufacturer,
    a.model,
    a.city,
    a.region,
    a.criticality,
    a.asset_status;


-- ============================================================
-- 4. Maintenance By Month
-- ============================================================

CREATE OR REPLACE VIEW motordatabricks.utilities_analytics.maintenance_by_month AS
SELECT
    d.year,
    d.month,
    d.month_name,
    COUNT(*) AS maintenance_count,
    ROUND(SUM(f.downtime_hours), 2) AS total_downtime_hours,
    ROUND(SUM(f.labor_hours), 2) AS total_labor_hours,
    ROUND(SUM(f.total_cost), 2) AS total_maintenance_cost,
    ROUND(AVG(f.total_cost), 2) AS avg_maintenance_cost
FROM motordatabricks.utilities_gold.fact_maintenance f
JOIN motordatabricks.utilities_gold.dim_date d
    ON f.created_date_key = d.date_key
GROUP BY
    d.year,
    d.month,
    d.month_name;


-- ============================================================
-- 5. Technician Performance
-- ============================================================

CREATE OR REPLACE VIEW motordatabricks.utilities_analytics.technician_performance AS
SELECT
    t.technician_id,
    t.technician_name,
    t.skill_level,
    t.team,
    t.experience_years,
    COUNT(*) AS maintenance_count,
    ROUND(SUM(f.labor_hours), 2) AS total_labor_hours,
    ROUND(AVG(f.labor_hours), 2) AS avg_labor_hours,
    ROUND(SUM(f.downtime_hours), 2) AS total_downtime_hours,
    ROUND(SUM(f.total_cost), 2) AS total_maintenance_cost,
    ROUND(AVG(f.total_cost), 2) AS avg_maintenance_cost
FROM motordatabricks.utilities_gold.fact_maintenance f
JOIN motordatabricks.utilities_gold.dim_technician t
    ON f.technician_key = t.technician_key
GROUP BY
    t.technician_id,
    t.technician_name,
    t.skill_level,
    t.team,
    t.experience_years;


-- ============================================================
-- 6. Maintenance Status Summary
-- ============================================================

CREATE OR REPLACE VIEW motordatabricks.utilities_analytics.maintenance_status_summary AS
SELECT
    s.work_order_status,
    s.resolution_code,
    COUNT(*) AS maintenance_count,
    ROUND(SUM(f.downtime_hours), 2) AS total_downtime_hours,
    ROUND(SUM(f.labor_hours), 2) AS total_labor_hours,
    ROUND(SUM(f.total_cost), 2) AS total_maintenance_cost,
    ROUND(AVG(f.total_cost), 2) AS avg_maintenance_cost
FROM motordatabricks.utilities_gold.fact_maintenance f
JOIN motordatabricks.utilities_gold.dim_status s
    ON f.status_key = s.status_key
GROUP BY
    s.work_order_status,
    s.resolution_code;
```
