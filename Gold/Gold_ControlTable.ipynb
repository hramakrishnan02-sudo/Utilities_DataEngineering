```python
%sql

CREATE TABLE IF NOT EXISTS motordatabricks.utilities_gold.cdf_control
(
    source_table STRING,
    last_processed_version BIGINT,
    updated_at TIMESTAMP
)
USING DELTA
LOCATION 'abfss://gold@utilitiesmaintainence.dfs.core.windows.net/cdf_control';
```

```python
%sql

DELETE FROM motordatabricks.utilities_gold.cdf_control;
```

```python
%sql

INSERT INTO motordatabricks.utilities_gold.cdf_control
SELECT
    'motordatabricks.utilities_silver.assets',
    MAX(version),
    current_timestamp()
FROM (
    DESCRIBE HISTORY motordatabricks.utilities_silver.assets
)

UNION ALL

SELECT
    'motordatabricks.utilities_silver.technicians',
    MAX(version),
    current_timestamp()
FROM (
    DESCRIBE HISTORY motordatabricks.utilities_silver.technicians
)

UNION ALL

SELECT
    'motordatabricks.utilities_silver.locations',
    MAX(version),
    current_timestamp()
FROM (
    DESCRIBE HISTORY motordatabricks.utilities_silver.locations
)

UNION ALL

SELECT
    'motordatabricks.utilities_silver.maintenance',
    MAX(version),
    current_timestamp()
FROM (
    DESCRIBE HISTORY motordatabricks.utilities_silver.maintenance
);
```
